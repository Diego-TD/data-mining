```python
import random
import numpy as np
import pandas as pd

random_list = [random.random() for _ in range(5_000_000)]
print("Random list length:", len(random_list))
```

```python
# population mean - for loop
sd = 0
n= 5_000_000
mean = 0
for number in random_list:
    mean += number
mean = mean/n

# sumatoria
sumatoria = 0
for number in random_list:
    sumatoria = (number - mean)**2

# divide by N
st = sumatoria / n

# sqrt
st = pow(sumatoria,0.5)

print("Standard deviation: ", st)
```

```python

list = np.array(random_list)

sd = np.std(list)
print("Standard deviation",sd)
```
